```python
import pandas as pd

df2 = pd.DataFrame(
    [["Mango", 50, 100],
     ["Guva", 60, 2],
     ["Kiwi", 50, 5]],
    columns=["Fruits", "Price", "Quantity"]
)

df2
```

```python
# Covert list into tuple

import pandas as pd

df2 = pd.DataFrame(
    (
        ("Mango", 50, 100),
        ("Guva", 60, 2),
        ("Kiwi", 50, 5)
    ),
    columns=("Fruits", "Price", "Quantity")
)

df2
```

```python
# Convert dictionary  to dataFrame

fav_hero = {
    "Fruits_names" : ["Mango","Kiwi","Guva"],
    "fee" : [30,40,50,]
}
df4 = pd.DataFrame(fav_hero)
df4
```

# Understanding the dataset using pandas

```python
df = pd.read_excel("Excel_data.xlsx")
```

```python
#Shape keyword will return "rows" & "columns"
df.shape
```

```python
# head() function returns top 5 records in pandas by default
df.head(10)
```

```python
# tail returns the bottom 5 records
df.tail(20)
```

```python
# Query : print Name column only
df.Name
```

```python
df.Age
```

```python
df['Age']
```

```python
# Print multiple columns  (Age and Salary)
df[['Age','salary','Profession']]
```

```python
df
```

```python
# Rename the columns

df2 = df.rename(columns  = {'Name' : 'Full Name','Age' : 'age','Profession' : 'Occupation'})
df2
```

```python
# Add a new column in dataset
# Rank func provide the rank based upon the salary
# Lower salary = Lower Rank

df['rank'] = df.salary.rank(ascending = False)
df.sort_values('rank')


# if column already present then this will update the column
# if column don't exist then new column will be created
```

```python
# Replace the names

df.Name = df.Name.replace({'Olivia' : 'Samantha'})
df
```

```python
# replace the multiple values

df.Name = df.Name.replace({"Samantha" : "Tamana","jake" : "BalKrishna"})
df
```

```python
# df.types ==> all the columns datatype

df.dtypes
```

```python
# convert one data type to another ===> astype

df.Name = df.Name.astype('string')
df.dtypes
df.salary = df.salary.astype('float')
df.dtypes
```

```python
# column names

df.columns

```

```python
# print only categorical column names

df.select_dtypes(include = ['str', 'string']).columns
```

```python
df.select_dtypes(include = ['int', 'float']).columns
```
