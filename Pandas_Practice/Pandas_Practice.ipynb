```python
import pandas as pd

df = pd.read_excel("Movies_Dataset.xlsx")

print(df.head())
```

```python
# 1. Select all movies in the "Action" genre

action_movies = df[df['Genre'] == 'Action']

print(action_movies)

# Explanation: Filters all rows where Genre is Action.
```

```python
# 2. Select movies released after 2000

movies_after_2000 = df[df['Release_Year'] > 2000]

print(movies_after_2000)
```

```python
# 3. Select movies with an IMDb rating greater than 8.5

high_rated = df[df['IMDb_Rating'] > 8.5]

print(high_rated)
```

```python
# 4. Select only "Movie Name" and "Genre" columns

selected_movies = df[['Movie Name', 'Genre']]

print(selected_movies)

# Explanation: Selects only the two required columns.
```

```python
# 5. Select movies released between 1990 and 2010. 

movies_between = df[
    (df['Release_Year'] >= 1990) &
    (df['Release_Year'] <= 2010)
]

print(movies_between)
```

```python
# 6. Count the number of movies in each genre. 

genre_count = df['Genre'].value_counts()

print(genre_count)
```

```python
# 7. Find the highest IMDb-rated movie

highest_rated = df[
    df['IMDb_Rating'] == df['IMDb_Rating'].max()
]

print(highest_rated)

# Explanation: max() finds the highest rating.
```

```python
# 8. Find the average IMDb rating of all movies

average_rating = df['IMDb_Rating'].mean()

print(average_rating)
```

```python
# 9. Find the total box office collection of all movies

total_revenue = df['Revenue'].sum()

print(total_revenue)
```

```python
# 10. Select movies with box office earnings greater than 500 million

high_revenue = df[df['Revenue'] > 500]

print(high_revenue)
```

```python
# 11. Rank movies based on IMDb rating

df['Rank'] = df['IMDb_Rating'].rank(
    ascending=False,
    method='dense'
)

print(df[['Movie Name', 'IMDb_Rating', 'Rank']])
```

```python
# 12. Sort movies by release year in descending order

duplicates = df[
    df['Movie Name'].duplicated(keep=False)
]

print(duplicates)
```

```python
# 13. Find all duplicate movie names (if any)

duplicates = df[
    df['Movie Name'].duplicated(keep=False)
]

print(duplicates)
```

```python
# 14. Remove duplicate movie names

df = df.drop_duplicates(
    subset='Movie Name',
    keep='first'
)

print(df)
```

```python
# 15. Replace "Sci-Fi" genre with "Science Fiction"

df['Genre'] = df['Genre'].replace(
    'Sci-Fi',
    'Science Fiction'
)

print(df)
```
