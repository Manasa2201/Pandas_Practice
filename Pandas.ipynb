# Grouping of data

```python
import pandas as pd
df = pd.read_excel("Excel_data.xlsx")


# group the data based upon profession
df.groupby("Profession").count()
```

```python
# find average salary based upon the profession

df.groupby("Profession").salary.mean()
```

```python
# group the data based upon multiple columns

df.groupby(["Profession", "Age", "Marriage_proposal_count" ]).salary.mean()
```

```python
# agg() : this function is used to apply multiple aggregate functions together

df[["Age", "salary"]].agg(["mean", "sum", "min", "max"])
```

```python
# apply different aggregate functions to different columns
df.agg({
    "Age" : ["min", "mean"],
    "salary" : ["max", "mean"]
})
```

```python
# agg() + groupby()
df.groupby("Profession").agg({
    "Age" : ["min", "mean"],
    "salary" : ["max", "mean"]
})
```

# Merging of dataframes

```python
# merge() : this function can be used to merge the dataframes based upon common column
# merge() function work like join in sql : inner, left, right outer

df1 = pd.DataFrame({'ID' : [1,2,3], 'Dream_company' : ['google', 'microsoft' ,'apple']})
df2 = pd.DataFrame({'ID' : [2,3,4], 'DreamPackage' : [50, 60, 40]})

merge_df = pd.merge(df1, df2, on = 'ID', how = 'inner')  # inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = 'ID', how = 'left')  # inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = 'ID', how = 'right')  # inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = 'ID', how = 'outer')  # inner join
merge_df
```

# join() : mainly used to work with index

```python
df3 = pd.DataFrame({"Hero_Name" : ["NTR", "Prabhas" , "Pawan"]}, index = [1,2,3])
df4 = pd.DataFrame({'Heronie_Name' : ["Kajol", "Anushka", "Shruti"]}, index = [2,3,4])

joined_df = df3.join(df4, how = 'inner')
joined_df
```

```python
joined_df = df3.join(df4, how = 'left')
joined_df
```

```python
joined_df = df3.join(df4, how = 'right')
joined_df
```

```python
joined_df = df3.join(df4, how = 'outer')
joined_df
```

# concat() 

```python
# Def: concat() function is used to stack dataframes vertically or horizontally
```

```python
df5 = pd.DataFrame({'ID' : [1,2], 'Fav_Movie' : ["Paradise", "Bahubali"]})
df6 = pd.DataFrame({'ID' : [3,4], 'Fav_Movie' : ["Spiderman", "Doomsday"]})
df7 = pd.DataFrame({'ID' : [5,6], 'Fav_Movie' : ["Puspa", "Magadeera"]})

# axis = 0 : Vertically (adding rows)
# axis = 1 : Horizontally (adding column)

concat_df = pd.concat([df5, df6, df7], axis = 0, ignore_index = True)
concat_df
```
